**1.** Дан случайный массив, поменять знак у элементов, значения которых между 3 и 8

```python
array = np.array([1, 5, 10, 3, 8, 2, 7, 12])
condition = (array >= 3) & (array <= 8)
array[condition] *= -1
print(array)
```

**2.** Заменить максимальный элемент случайного массива на 0

```python
array = np.array([1, 5, 10, 3, 8, 2, 7, 12])
array[array.argmax()] = 0
print(array)
```

**3.** Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

```python
array = np.array([
    [1, 2, 3, 4],
    [0, 1, 2, 3]
    ])
result = np.outer(array.flatten(), array.flatten())
print(result)
```

* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
x = np.array([[1, 0, 1], 
              [2, 0, 2], 
              [3, 0, 3], 
              [4, 4, 4]])
diag = []
counter = 0
for i in x:
    if len(i) <= counter:
        break
    diag.append(i[counter])
    counter += 1
no_zeros = [i for i in diag if i != 0]
result = 1
for i in no_zeros:
    result *= i
print(result)

#numpy method

print(np.prod([np.diag(x)[np.diag(x) != 0]]))
```

* __Задача 2__: Даны два вектора x и y. Проверить, задают ли они одно и то же мультимножество.

```python
x = np.array([1, 2, 2, 4])
y = np.array([4, 2, 1, 2])

x_dict = {}
for i in x:
    if i not in x_dict:
        x_dict[i] = 1
    else:
        x_dict[i] += 1
y_dict = {}
for i in y:
    if i not in y_dict:
        y_dict[i] = 1
    else:
        y_dict[i] += 1
x_dict == y_dict


#numpy method

len(x) == len(y) and np.all(np.sort(x) == np.sort(y))
 
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль. 
 Например, для x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) ответ 5.

```python
a = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
largest = 0
previous_zero = False
for i in a:
    if previous_zero:
        if i > largest:
            largest = i
    if i == 0:
        previous_zero = True
    else:
        previous_zero = False

#numpy method

largest = np.max(a[1:][a[:-1] == 0])
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = np.array([2, 2, 2, 3, 3, 3, 5])
x_dict = {}
for i in x:
    if i not in x_dict:
        x_dict[i] = 1
    else:
        x_dict[i] += 1
keys = []
values = []
for i, j in x_dict.items():
    keys.append(int(i))
    values.append(j)
print(f'{keys}, {values}')

#numpy method

np.unique(x, return_counts=True)
```
